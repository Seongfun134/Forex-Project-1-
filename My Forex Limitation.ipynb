## Forex Project Explaination 

## My Limitations

- My dataset only has daily and weekly data, so I could not test trading sessions (Asia, London, New York) or trades per day.
- I only backtested 5 simple trading strategies. No complex strategies were tested.
- My data ends in July 2025, so newer market data is not included.
- Some trading pair like CNY/USD are pegged ,meaning the price mostly cannot move , this make backtesting cannot backtest properly 
- This forex project don't have include spread and commission , but in actual forex trading the result would be lower.
- The data shows past results only. It mean the strategy can work in the pass may not work in the future

## Why I want to create this project for my future use

- When I learn a new trading strategy, I can test it first to see whether it works before using real money.
- It saves time compared to backtesting manually on a chart.
- It shows the win rate and profit factor over the past 25 years, so I can decide if the strategy is good to use or not.
- It shows how much profit or loss the strategy would have made over the past 25 years.
